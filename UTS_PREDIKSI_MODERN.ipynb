# EXPERIMENT 1

STEP 1.1 — Import Model dan Metrik

```python
# Import Logistic Regression
from sklearn.linear_model import LogisticRegression

# Import evaluasi
from sklearn.model_selection import cross_val_score, StratifiedKFold
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score
)
```

STEP 1.2 — Membuat Model Baseline

```python
# Membuat model Logistic Regression baseline
logreg_baseline = LogisticRegression(
    max_iter=1000,
    random_state=42
)
```

STEP 1.3 — 5-Fold Cross Validation

```python
# Membuat 5-fold stratified cross validation
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# Menghitung F1-score pada setiap fold
cv_f1_scores = cross_val_score(
    logreg_baseline,
    X_train,
    y_train,
    cv=cv,
    scoring='f1'
)

print("F1-score setiap fold:")
for i, score in enumerate(cv_f1_scores, start=1):
    print(f"Fold {i}: {score:.4f}")

print(f"\nMean CV F1-score: {cv_f1_scores.mean():.4f}")
print(f"Std CV F1-score : {cv_f1_scores.std():.4f}")
```

STEP 1.4 — Training Model

```python
# Training Logistic Regression pada seluruh data training
logreg_baseline.fit(X_train, y_train)
```

STEP 1.5 — Prediksi Data Test

```python
# Prediksi data test
y_pred_baseline = logreg_baseline.predict(X_test)
```

STEP 1.6 — Menghitung Metrik Test

```python
# Menghitung metrik evaluasi
accuracy_baseline = accuracy_score(y_test, y_pred_baseline)
precision_baseline = precision_score(y_test, y_pred_baseline)
recall_baseline = recall_score(y_test, y_pred_baseline)
f1_baseline = f1_score(y_test, y_pred_baseline)

print("=== HASIL EVALUASI LOGISTIC REGRESSION BASELINE ===")
print(f"Accuracy : {accuracy_baseline:.4f}")
print(f"Precision: {precision_baseline:.4f}")
print(f"Recall   : {recall_baseline:.4f}")
print(f"F1-score : {f1_baseline:.4f}")
print(f"CV F1-score: {cv_f1_scores.mean():.4f}")
```

# EXPERIMENT 2

STEP 2.1 — Import StandardScaler dan Pipeline

```python
# Import preprocessing dan pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
```

STEP 2.2 — Membuat Pipeline

```python
# Membuat pipeline StandardScaler + Logistic Regression
logreg_preprocessed = Pipeline([
    ('scaler', StandardScaler()),
    ('logreg', LogisticRegression(
        max_iter=1000,
        random_state=42
    ))
])
```

STEP 2.3 — 5-Fold Cross Validation

```python
# 5-fold Cross Validation untuk model preprocessing
cv_f1_scores_preprocessed = cross_val_score(
    logreg_preprocessed,
    X_train,
    y_train,
    cv=cv,
    scoring='f1'
)

print("F1-score setiap fold:")
for i, score in enumerate(cv_f1_scores_preprocessed, start=1):
    print(f"Fold {i}: {score:.4f}")

print(f"\nMean CV F1-score: {cv_f1_scores_preprocessed.mean():.4f}")
print(f"Std CV F1-score : {cv_f1_scores_preprocessed.std():.4f}")
```

STEP 2.4 — Training

```python
# Training model
logreg_preprocessed.fit(X_train, y_train)
```

STEP 2.5 — Prediksi

```python
# Prediksi data test
y_pred_preprocessed = logreg_preprocessed.predict(X_test)
```

STEP 2.6 — Evaluasi

```python
# Menghitung metrik
accuracy_preprocessed = accuracy_score(
    y_test,
    y_pred_preprocessed
)

precision_preprocessed = precision_score(
    y_test,
    y_pred_preprocessed
)

recall_preprocessed = recall_score(
    y_test,
    y_pred_preprocessed
)

f1_preprocessed = f1_score(
    y_test,
    y_pred_preprocessed
)

print("=== HASIL EVALUASI LOGISTIC REGRESSION + PREPROCESSING ===")
print(f"Accuracy   : {accuracy_preprocessed:.4f}")
print(f"Precision  : {precision_preprocessed:.4f}")
print(f"Recall     : {recall_preprocessed:.4f}")
print(f"F1-score   : {f1_preprocessed:.4f}")
print(f"CV F1-score: {cv_f1_scores_preprocessed.mean():.4f}")
```

STEP 2.7 — Membandingkan Experiment 1 vs Experiment 2

```python
comparison_exp1_exp2 = pd.DataFrame({
    'Metric': [
        'CV F1-score',
        'Test Accuracy',
        'Test Precision',
        'Test Recall',
        'Test F1-score'
    ],
    'Experiment 1 - Baseline': [
        cv_f1_scores.mean(),
        accuracy_baseline,
        precision_baseline,
        recall_baseline,
        f1_baseline
    ],
    'Experiment 2 - Preprocessing': [
        cv_f1_scores_preprocessed.mean(),
        accuracy_preprocessed,
        precision_preprocessed,
        recall_preprocessed,
        f1_preprocessed
    ]
})

comparison_exp1_exp2
```

# EXPERIMENT 3 — PERBANDINGAN MODEL

STEP 3.1 — Import Model

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
```

STEP 3.2 — Membuat Model

```python
# Logistic Regression + StandardScaler
model_logreg = logreg_preprocessed

# Decision Tree
model_dt = DecisionTreeClassifier(
    random_state=42
)

# Random Forest
model_rf = RandomForestClassifier(
    random_state=42
)
```

STEP 3.3 — 5-Fold Cross Validation

```python
# Logistic Regression
cv_f1_logreg = cross_val_score(
    model_logreg,
    X_train,
    y_train,
    cv=cv,
    scoring='f1'
)

# Decision Tree
cv_f1_dt = cross_val_score(
    model_dt,
    X_train,
    y_train,
    cv=cv,
    scoring='f1'
)

# Random Forest
cv_f1_rf = cross_val_score(
    model_rf,
    X_train,
    y_train,
    cv=cv,
    scoring='f1'
)

# Menampilkan hasil
print("=== CV F1-SCORE ===")

print("\nLogistic Regression:")
for i, score in enumerate(cv_f1_logreg, start=1):
    print(f"Fold {i}: {score:.4f}")
print(f"Mean CV F1: {cv_f1_logreg.mean():.4f}")
print(f"Std CV F1 : {cv_f1_logreg.std():.4f}")

print("\nDecision Tree:")
for i, score in enumerate(cv_f1_dt, start=1):
    print(f"Fold {i}: {score:.4f}")
print(f"Mean CV F1: {cv_f1_dt.mean():.4f}")
print(f"Std CV F1 : {cv_f1_dt.std():.4f}")

print("\nRandom Forest:")
for i, score in enumerate(cv_f1_rf, start=1):
    print(f"Fold {i}: {score:.4f}")
print(f"Mean CV F1: {cv_f1_rf.mean():.4f}")
print(f"Std CV F1 : {cv_f1_rf.std():.4f}")
```

STEP 3.4 — Training Ketiga Model

```python
# Training Logistic Regression
model_logreg.fit(X_train, y_train)

# Training Decision Tree
model_dt.fit(X_train, y_train)

# Training Random Forest
model_rf.fit(X_train, y_train)
```

STEP 3.5 — Prediksi Data Test

```python
y_pred_logreg = model_logreg.predict(X_test)
y_pred_dt = model_dt.predict(X_test)
y_pred_rf = model_rf.predict(X_test)
```

STEP 3.6 — Evaluasi Ketiga Model

```python
# Logistic Regression
accuracy_logreg = accuracy_score(y_test, y_pred_logreg)
precision_logreg = precision_score(y_test, y_pred_logreg)
recall_logreg = recall_score(y_test, y_pred_logreg)
f1_logreg = f1_score(y_test, y_pred_logreg)

# Decision Tree
accuracy_dt = accuracy_score(y_test, y_pred_dt)
precision_dt = precision_score(y_test, y_pred_dt)
recall_dt = recall_score(y_test, y_pred_dt)
f1_dt = f1_score(y_test, y_pred_dt)

# Random Forest
accuracy_rf = accuracy_score(y_test, y_pred_rf)
precision_rf = precision_score(y_test, y_pred_rf)
recall_rf = recall_score(y_test, y_pred_rf)
f1_rf = f1_score(y_test, y_pred_rf)

print("=== HASIL EVALUASI DATA TEST ===")

print("\nLogistic Regression + Scaling")
print(f"Accuracy : {accuracy_logreg:.4f}")
print(f"Precision: {precision_logreg:.4f}")
print(f"Recall   : {recall_logreg:.4f}")
print(f"F1-score : {f1_logreg:.4f}")

print("\nDecision Tree")
print(f"Accuracy : {accuracy_dt:.4f}")
print(f"Precision: {precision_dt:.4f}")
print(f"Recall   : {recall_dt:.4f}")
print(f"F1-score : {f1_dt:.4f}")

print("\nRandom Forest")
print(f"Accuracy : {accuracy_rf:.4f}")
print(f"Precision: {precision_rf:.4f}")
print(f"Recall   : {recall_rf:.4f}")
print(f"F1-score : {f1_rf:.4f}")
```

# EXPERIMENT 4 — GridSearchCV

STEP 4.1 — Import GridSearchCV

```python
from sklearn.model_selection import GridSearchCV
```

STEP 4.2 — Menentukan Parameter Grid

```python
param_grid_rf = {
    'n_estimators': [100],
    'max_depth': [None, 10],
    'min_samples_split': [2, 5]
}
```

STEP 4.3 — GridSearchCV

```python
grid_search_rf = GridSearchCV(
    estimator=RandomForestClassifier(
        random_state=42
    ),
    param_grid=param_grid_rf,
    scoring='f1',
    cv=cv,
    n_jobs=-1,
    verbose=1
)

grid_search_rf.fit(X_train, y_train)
```

STEP 4.4 — Melihat Parameter Terbaik

```python
print("=== HASIL GRID SEARCH RANDOM FOREST ===")
print("\nBest Parameters:")
print(grid_search_rf.best_params_)

print(f"\nBest CV F1-score: {grid_search_rf.best_score_:.4f}")
```

Step 4.5 — Evaluasi Random Forest setelah tuning

```python
# Mengambil model terbaik hasil Grid Search
best_rf = grid_search_rf.best_estimator_

# Prediksi pada data testing
y_pred_rf_tuned = best_rf.predict(X_test)

# Menghitung metrik evaluasi
accuracy_rf_tuned = accuracy_score(y_test, y_pred_rf_tuned)
precision_rf_tuned = precision_score(y_test, y_pred_rf_tuned)
recall_rf_tuned = recall_score(y_test, y_pred_rf_tuned)
f1_rf_tuned = f1_score(y_test, y_pred_rf_tuned)

print("=== HASIL EVALUASI RANDOM FOREST SETELAH TUNING ===")
print(f"Accuracy  : {accuracy_rf_tuned:.4f}")
print(f"Precision : {precision_rf_tuned:.4f}")
print(f"Recall    : {recall_rf_tuned:.4f}")
print(f"F1-score  : {f1_rf_tuned:.4f}")
```
